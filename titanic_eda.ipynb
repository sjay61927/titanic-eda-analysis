```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')

# Start here — just run this first and read the output carefully
print(df.shape)
print(df.head())
print(df.info())
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print(df.isnull().sum())
print("\n--- Percentage missing ---")
print((df.isnull().sum() / len(df) * 100).round(2))
```

```python
# Q2: Overall survival rate
survival_rate = df['Survived'].mean() * 100
print(f"Overall survival rate: {survival_rate:.2f}%")
```

```python
# Q3: Survival rate by gender
print("Survival rate by gender:")
print(df.groupby('Sex')['Survived'].mean() * 100)

# Q4: Survival rate by passenger class
print("\nSurvival rate by passenger class:")
print(df.groupby('Pclass')['Survived'].mean() * 100)
```

```python
# Q5: Age distribution - survivors vs non-survivors
plt.figure(figsize=(10, 5))

sns.histplot(data=df, x='Age', hue='Survived',
             bins=30, kde=True,
             palette={0:'salmon', 1:'steelblue'})

plt.title('Age Distribution: Survivors vs Non-Survivors')
plt.xlabel('Age')
plt.ylabel('Count')
plt.legend(labels=['Did not survive', 'Survived'])
plt.show()
```

```python
# Q6: Average fare by survival
print("Average fare paid:")
print(df.groupby('Survived')['Fare'].mean().round(2))
```
